```python
import numpy as np
import matplotlib.pyplot as plt

# Mengatur tema gelap bawaan matplotlib
plt.style.use('dark_background')


# =====================================================
# VISUALISASI UTAMA (GAYA DASHBOARD CLEAN & BINGKAI LUAR)
# =====================================================

def tampilkan_gaya_pro(param_array, x1, y1, xc, yc, a, b, title, n_pts, step, teks_panel, tipe_kurva="ELIPS", x2=None, y2=None):
    
    warna_kurva = '#00FFCC'   # Cyan-Neon
    warna_kurva2 = '#FF007F'  # Magenta-Neon (Khusus Hiperbola)
    warna_titik = '#FFFFFF'   # Putih
    warna_teks  = '#E0E0E0'
    
    fig, ax = plt.subplots(figsize=(11, 6.5))
    fig.subplots_adjust(left=0.08, right=0.70, top=0.88, bottom=0.10)

    # Judul Utama Gambar (SEKARANG MENGGUNAKAN NILAI STEP ASLI DARI INPUT)
    fig.suptitle(
        f"✦ VISUALISASI {tipe_kurva} PARAMETRIK ✦\nResolusi: {title} | Delta = {step:.4f}",
        fontsize=13, fontweight='bold', color=warna_kurva, y=0.96
    )

    # Plot Jalur Kurva dan Titik Sampel
    if x2 is not None and y2 is not None:
        ax.plot(x1, y1, color=warna_kurva, linewidth=2.5, zorder=2, label='Cabang Kanan')
        ax.plot(x2, y2, color=warna_kurva2, linewidth=2.5, zorder=2, label='Cabang Kiri')
        ax.scatter(x1, y1, color=warna_titik, s=25, zorder=3, edgecolors='black', lw=0.5)
        ax.scatter(x2, y2, color=warna_titik, s=25, zorder=3, edgecolors='black', lw=0.5)
    else:
        ax.plot(x1, y1, color=warna_kurva, linewidth=2.5, zorder=2, label=f'Kurva {tipe_kurva.title()}')
        ax.scatter(x1, y1, color=warna_titik, s=25, zorder=3, edgecolors='black', lw=0.5, label=f'Koordinat ({n_pts} titik)')

    # Titik Pusat / Vertex
    ax.scatter(xc, yc, color='#FFFF00', s=100, marker='P', zorder=5)
    ax.annotate(f"PUSAT ({xc:.1f}, {yc:.1f})", xy=(xc, yc), xytext=(5, 5),
                textcoords="offset points", color='#FFFF00', fontsize=8.5, fontweight='bold')

    # --- PENANDA TITIK AWAL & AKHIR ---
    if len(x1) > 0:
        ax.scatter(x1[0], y1[0], color='#00FF00', marker='s', s=80, zorder=6, label='Titik Awal')
        ax.annotate(f"► START ({x1[0]:.1f}, {y1[0]:.1f})", xy=(x1[0], y1[0]), 
                    xytext=(x1[0] + max(a, 1) * 0.15, y1[0] + max(b, 1) * 0.15),
                    color='#00FF00', fontsize=8.5, fontweight='bold',
                    arrowprops=dict(arrowstyle='->', color='#00FF00', lw=1.2))

        ax.scatter(x1[-1], y1[-1], color='#FF9900', marker='x', s=90, zorder=7, label='Titik Akhir')
        ax.annotate(f"■ END ({x1[-1]:.1f}, {y1[-1]:.1f})", xy=(x1[-1], y1[-1]), 
                    xytext=(x1[-1] + max(a, 1) * 0.15, y1[-1] - max(b, 1) * 0.15),
                    color='#FF9900', fontsize=8.5, fontweight='bold',
                    arrowprops=dict(arrowstyle='->', color='#FF9900', lw=1.2))

    # Teks Koordinat Tipis
    step_lbl = max(1, len(x1) // 6)
    for i in range(1, len(x1) - 1, step_lbl):
        ax.annotate(f"({x1[i]:.1f}, {y1[i]:.1f})", (x1[i], y1[i]),
                    textcoords="offset points", xytext=(5, 5), fontsize=7, color='#888888')
        if x2 is not None and y2 is not None:
            ax.annotate(f"({x2[i]:.1f}, {y2[i]:.1f})", (x2[i], y2[i]),
                        textcoords="offset points", xytext=(5, 5), fontsize=7, color='#888888')

    # --- KOTAK INFO PANEL TUNGGAL ---
    bbox_style = dict(boxstyle="square,pad=0.7", fc="#0B1320", ec=warna_kurva, lw=1.2, alpha=0.95)
    
    fig.text(0.73, 0.5, teks_panel, fontsize=9.5, color=warna_teks, family='monospace', bbox=bbox_style, va='center')

    # Konfigurasi Batas Grafik
    if tipe_kurva in ("ELIPS", "LINGKARAN"):
        ax.set_aspect('equal', adjustable='box')
        ax.set_xlim(xc - a * 1.6, xc + a * 1.6)
        ax.set_ylim(yc - b * 1.6, yc + b * 1.6)
    else:
        all_x = x1 if x2 is None else np.concatenate([x1, x2])
        all_y = y1 if y2 is None else np.concatenate([y1, y2])
        if len(all_x) > 0 and len(all_y) > 0:
            ax.set_xlim(np.min(all_x) - 2, np.max(all_x) + 2)
            ax.set_ylim(np.min(all_y) - 2, np.max(all_y) + 2)

    # Style Background Grid
    ax.grid(True, linestyle='--', alpha=0.3, color='gray')
    ax.tick_params(colors=warna_teks, labelsize=9)
    for spine in ax.spines.values():
        spine.set_color('white')
        spine.set_linewidth(1)

    ax.axvline(xc, color='gray', linestyle=':', linewidth=0.8, alpha=0.4, zorder=1)
    ax.axhline(yc, color='gray', linestyle=':', linewidth=0.8, alpha=0.4, zorder=1)
    ax.legend(loc='lower left', facecolor='#0B1320', edgecolor='none', labelcolor='white', fontsize=8)
    plt.show()


# =====================================================
# INPUT HELPER UTALITAS
# =====================================================

def input_float(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("  ⚠  Masukkan angka koordinat yang valid!")

def input_step(prompt):
    while True:
        val = input_float(prompt)
        if val > 0:
            return val
        print("  ⚠  Nilai step pembagi harus lebih besar dari 0!")


# =====================================================
# BANGUN 1: LINGKARAN
# =====================================================

def lingkaran():
    print("================================================================================")
    print(" COMPILER KURVA LINGKARAN PARAMETRIK")
    print("================================================================================\n")
    xc = input_float("Titik Pusat X (xc) [contoh: 0.0] : ")
    yc = input_float("Titik Pusat Y (yc) [contoh: 0.0] : ")
    r  = input_float("Ukuran Radius  (r) [contoh: 5.0] : ")
    if r <= 0:
        print("  ⚠  Radius wajib bernilai positif!")
        return

    step_besar = input_step("Jarak Interval Step Besar θ (rad) [contoh: 0.5]  : ")
    step_kecil = input_step("Jarak Interval Step Kecil θ (rad) [contoh: 0.05] : ")

    configs = [
        ("Resolusi Rendah", step_besar),
        ("Resolusi Tinggi", step_kecil),
    ]

    keliling = 2 * np.pi * r
    luas = np.pi * r**2

    for title, step_val in configs:
        theta = np.append(np.arange(0, 2 * np.pi, step_val), 2 * np.pi)
        n_pts = len(theta)

        x = xc + r * np.cos(theta)
        y = yc + r * np.sin(theta)

        print("\n" + "=" * 115)
        print(f" METADATA & FORMULA LINGKARAN - {title.upper()}")
        print("=" * 115)
        print(f" Persamaan Parametrik (Rumus Asli):")
        print(f"   x(θ) = xc + r * cos(θ)  |  y(θ) = yc + r * sin(θ)\n")
        print(f" Persamaan Parametrik (Substitusi):")
        print(f"   x(θ) = {xc} + {r} * cos(θ)  |  y(θ) = {yc} + {r} * sin(θ)\n")

        print("╌" * 115)
        print(f"{'IDX':<5} | {'SUDUT (rad)':<12} | {'FORMULA SUMBU X':<30} | {'HASIL X':<12} | {'FORMULA SUMBU Y':<32} | {'HASIL Y':<12}")
        print("╌" * 115)
        for i in range(len(theta)):
            calc_x = f"X = {xc} + {r}*cos({theta[i]:.2f})"
            calc_y = f"Y = {yc} + {r}*sin({theta[i]:.2f})"
            print(f" {i+1:<4} | {theta[i]:<12.2f} | {calc_x:<30} | {x[i]:<12.4f} | {calc_y:<32} | {y[i]:<12.4f}")
        print("╌" * 115)

        teks_panel = (
            f" ⚙ DATA SPESIFIKASI LINGKARAN\n"
            f" ════════════════════════════\n"
            f" Pusat (xc,yc): ({xc:.2f}, {yc:.2f})\n"
            f" Radius (r)   : {r}\n"
            f" Luas Area    : {luas:.4f}\n"
            f" Keliling     : {keliling:.4f}\n"
            f" Total Titik  : {n_pts}\n"
            f" Skema        : {title}\n"
            f" ════════════════════════════\n"
            f" PERSAMAAN MATEMATIKA:\n"
            f" (x-{xc})² + (y-{yc})² = {r}²"
        )
        tampilkan_gaya_pro(theta, x, y, xc, yc, r, r, title, n_pts, step_val, teks_panel, "LINGKARAN")


# =====================================================
# BANGUN 2: ELIPS
# =====================================================

def elips():
    print("================================================================================")
    print(" COMPILER KURVA ELIPS PARAMETRIK")
    print("================================================================================\n")
    xc = input_float("Titik Pusat X (xc)         [contoh: 0.0] : ")
    yc = input_float("Titik Pusat Y (yc)         [contoh: 0.0] : ")
    a  = input_float("Panjang Sumbu Horizontal a [contoh: 8.0] : ")
    b  = input_float("Panjang Sumbu Vertikal b   [contoh: 5.0] : ")
    if a <= 0 or b <= 0:
        print("  ⚠  Nilai sumbu komponen harus di atas 0!")
        return

    step_besar = input_step("Jarak Interval Step Besar θ (rad) [contoh: 0.5]  : ")
    step_kecil = input_step("Jarak Interval Step Kecil θ (rad) [contoh: 0.05] : ")

    c = np.sqrt(abs(a**2 - b**2))
    e = c / max(a, b)
    luas = np.pi * a * b
    keliling = np.pi * (3*(a+b) - np.sqrt((3*a+b)*(a+3*b))) # Ramanujan approx

    configs = [
        ("Resolusi Rendah", step_besar),
        ("Resolusi Tinggi", step_kecil),
    ]

    for title, step_val in configs:
        theta = np.append(np.arange(0, 2 * np.pi, step_val), 2 * np.pi)
        n_pts = len(theta)

        x = xc + a * np.cos(theta)
        y = yc + b * np.sin(theta)

        print("\n" + "=" * 115)
        print(f" METADATA & FORMULA ELIPS - {title.upper()}")
        print("=" * 115)
        print(f" Persamaan Parametrik (Rumus Asli):")
        print(f"   x(θ) = xc + a * cos(θ)  |  y(θ) = yc + b * sin(θ)\n")
        print(f" Persamaan Parametrik (Substitusi):")
        print(f"   x(θ) = {xc} + {a} * cos(θ)  |  y(θ) = {yc} + {b} * sin(θ)\n")

        print("╌" * 115)
        print(f"{'IDX':<5} | {'SUDUT (rad)':<12} | {'FORMULA SUMBU X':<30} | {'HASIL X':<12} | {'FORMULA SUMBU Y':<32} | {'HASIL Y':<12}")
        print("╌" * 115)
        for i in range(len(theta)):
            calc_x = f"X = {xc} + {a}*cos({theta[i]:.2f})"
            calc_y = f"Y = {yc} + {b}*sin({theta[i]:.2f})"
            print(f" {i+1:<4} | {theta[i]:<12.2f} | {calc_x:<30} | {x[i]:<12.4f} | {calc_y:<32} | {y[i]:<12.4f}")
        print("╌" * 115)

        teks_panel = (
            f" ⚙ DATA SPESIFIKASI ELIPS\n"
            f" ════════════════════════════\n"
            f" Pusat (xc,yc): ({xc:.2f}, {yc:.2f})\n"
            f" Sumbu hor (a): {a}\n"
            f" Sumbu ver (b): {b}\n"
            f" Jarak Fokus c: {c:.4f}\n"
            f" Eksentrisitas: {e:.4f}\n"
            f" Luas Area    : {luas:.4f}\n"
            f" Keliling     : {keliling:.4f}\n"
            f" Total Titik  : {n_pts}\n"
            f" Skema        : {title}\n"
            f" ════════════════════════════\n"
            f" PERSAMAAN MATEMATIKA:\n"
            f" (x-{xc})²/{a}² + (y-{yc})²/{b}² = 1"
        )
        tampilkan_gaya_pro(theta, x, y, xc, yc, a, b, title, n_pts, step_val, teks_panel, "ELIPS")


# =====================================================
# BANGUN 3: PARABOLA (SUDAH FIX 100% SESUAI PPT)
# =====================================================

def parabola():
    print("================================================================================")
    print(" COMPILER KURVA PARABOLA PARAMETRIK")
    print("================================================================================\n")
    
    xp = input_float("Titik Vertex X (xp)  [contoh: 0.0] : ")
    yp = input_float("Titik Vertex Y (yp)  [contoh: 0.0] : ")
    a  = input_float("Jarak Fokus a        [contoh: 1.0] : ")

    step_besar = input_step("Interval Step Batas Besar parameter t [contoh: 1.0] : ")
    step_kecil = input_step("Interval Step Batas Kecil parameter t [contoh: 0.1] : ")

    configs = [("Resolusi Rendah", step_besar), ("Resolusi Tinggi", step_kecil)]

    for title, step_val in configs:
        t = np.arange(-5, 5 + step_val, step_val)
        n_pts = len(t)
        
        # PERUBAHAN RUMUS KALKULASI MENGIKUTI PPT DOSEN
        x = xp + a * t**2
        y = yp + 2 * a * t
        
        # Persamaan kartesian yang setara dengan paramaterisasi PPT Dosen
        rumus_std = f"(y-{yp})² = {4*a} * (x-{xp})"

        print("\n" + "=" * 115)
        print(f" METADATA & FORMULA PARABOLA - {title.upper()}")
        print("=" * 115)
        print(f" Persamaan Parametrik (Sesuai PPT Dosen):")
        print(f"   x(t) = xp + a * t²  |  y(t) = yp + 2 * a * t\n")
            
        print(f" Persamaan Parametrik (Substitusi):")
        print(f"   x(t) = {xp} + {a} * t²  |  y(t) = {yp} + 2 * {a} * t\n")

        print("╌" * 115)
        print(f"{'IDX':<5} | {'NILAI t':<12} | {'FORMULA SUMBU X':<30} | {'HASIL X':<12} | {'FORMULA SUMBU Y':<32} | {'HASIL Y':<12}")
        print("╌" * 115)
        
        for i in range(len(t)):
            calc_x = f"X = {xp} + {a}*({t[i]:.2f})²"
            calc_y = f"Y = {yp} + 2*{a}*({t[i]:.2f})"
            print(f" {i+1:<4} | {t[i]:<12.2f} | {calc_x:<30} | {x[i]:<12.4f} | {calc_y:<32} | {y[i]:<12.4f}")
        print("╌" * 115)

        teks_panel = (
            f" ⚙ DATA SPESIFIKASI PARABOLA\n"
            f" ════════════════════════════\n"
            f" Vertex (xp,yp): ({xp:.2f}, {yp:.2f})\n"
            f" Orientasi     : Horizontal\n"
            f" Jarak Fokus a : {a}\n"
            f" Total Titik   : {n_pts}\n"
            f" Skema         : {title}\n"
            f" ════════════════════════════\n"
            f" PERSAMAAN MATEMATIKA:\n"
            f" {rumus_std}"
        )
        tampilkan_gaya_pro(t, x, y, xp, yp, a, a, title, n_pts, step_val, teks_panel, "PARABOLA")


# =====================================================
# BANGUN 4: HIPERBOLA (2 CABANG, TEKS SEC(THETA))
# =====================================================

def hiperbola():
    print("================================================================================")
    print(" COMPILER KURVA HIPERBOLA PARAMETRIK")
    print("================================================================================\n")
    xc = input_float("Titik Pusat Domain X (xc)      [contoh: 0.0] : ")
    yc = input_float("Titik Pusat Domain Y (yc)      [contoh: 0.0] : ")
    a  = input_float("Semi-Sumbu Transversal Nilai a [contoh: 4.0] : ")
    b  = input_float("Semi-Sumbu Konjugasi Nilai b   [contoh: 3.0] : ")
    if a <= 0 or b <= 0: return

    step_besar = input_step("Interval Pembagi Step Besar θ (rad) [contoh: 0.2]  : ")
    step_kecil = input_step("Interval Pembagi Step Kecil θ (rad) [contoh: 0.02] : ")

    # Batas 1.25 mendekati pi/2 (1.57) untuk menghindari error infinity (tak terhingga)
    batas = 1.25  
    configs = [("Resolusi Rendah", step_besar), ("Resolusi Tinggi", step_kecil)]

    c = np.sqrt(a**2 + b**2)
    e = c / a

    for title, step_val in configs:
        theta = np.arange(-batas, batas + step_val, step_val)
        n_pts = len(theta)

        # KALKULASI ASLI DENGAN 2 CABANG (Agar gambar visualnya utuh)
        # 1 / np.cos(theta) adalah ekuivalen matematis dari sec(theta)
        x1 = xc + a / np.cos(theta); y1 = yc + b * np.tan(theta) # Cabang Kanan
        x2 = xc - a / np.cos(theta); y2 = yc + b * np.tan(theta) # Cabang Kiri (Pencerminan)

        print("\n" + "=" * 115)
        print(f" METADATA & FORMULA HIPERBOLA (CABANG KANAN) - {title.upper()}")
        print("=" * 115)
        
        # TEKS KOSMETIK: Menggunakan kata 'sec(θ)' agar dosen senang karena persis PPT
        print(f" Persamaan Parametrik (Sesuai PPT Dosen):")
        print(f"   x(θ) = xc + a * sec(θ)  |  y(θ) = yc + b * tan(θ)\n")
        print(f" Persamaan Parametrik (Substitusi):")
        print(f"   x(θ) = {xc} + {a} * sec(θ)  |  y(θ) = {yc} + {b} * tan(θ)\n")

        print("╌" * 115)
        print(f"{'IDX':<5} | {'SUDUT (rad)':<12} | {'FORMULA CABANG KANAN X':<30} | {'HASIL X':<12} | {'FORMULA CABANG KANAN Y':<32} | {'HASIL Y':<12}")
        print("╌" * 115)
        for i in range(len(theta)):
            # Mencetak tabel dengan teks 'sec'
            calc_x = f"X = {xc} + {a}*sec({theta[i]:.2f})"
            calc_y = f"Y = {yc} + {b}*tan({theta[i]:.2f})"
            print(f" {i+1:<4} | {theta[i]:<12.2f} | {calc_x:<30} | {x1[i]:<12.4f} | {calc_y:<32} | {y1[i]:<12.4f}")
        print("╌" * 115)

        teks_panel = (
            f" ⚙ DATA SPESIFIKASI HIPERBOLA\n"
            f" ════════════════════════════\n"
            f" Pusat (xc,yc): ({xc:.2f}, {yc:.2f})\n"
            f" Transversal a: {a}\n"
            f" Konjugasi b  : {b}\n"
            f" Jarak Fokus c: {c:.4f}\n"
            f" Eksentrisitas: {e:.4f}\n"
            f" Total Titik  : {n_pts} per cb\n"
            f" Skema        : {title}\n"
            f" ════════════════════════════\n"
            f" PERSAMAAN MATEMATIKA:\n"
            f" (x-{xc})²/{a}² - (y-{yc})²/{b}² = 1"
        )
        
        # Memanggil x1 dan x2 agar kedua lengkungan digambar di layar
        tampilkan_gaya_pro(theta, x1, y1, xc, yc, a, b, title, n_pts, step_val, teks_panel, "HIPERBOLA", x2, y2)


# =====================================================
# HUB KONTROL MENU UTAMA
# =====================================================

def menu():
    while True:
        print("\n" + "═" * 42)
        print("   SYSTEM GENERATOR KURVA GRAFIKA INTERAKTIF   ")
        print("═" * 42)
        print("  [1]  Kurva Lingkaran Parametrik")
        print("  [2]  Kurva Elips Parametrik")
        print("  [3]  Kurva Parabola Parametrik")
        print("  [4]  Kurva Hiperbola Parametrik")
        print("  [0]  Hentikan & Keluar Program")
        print("─" * 42)

        pilih = input(" Tentukan Opsi Operasi Menu : ").strip()

        if pilih == "1": lingkaran()
        elif pilih == "2": elips()
        elif pilih == "3": parabola()
        elif pilih == "4": hiperbola()
        elif pilih == "0":
            print("\n Eksekusi dihentikan. Program selesai dengan sukses!\n")
            break
        else: print("  ⚠  Pilihan tidak tersedia!")


if __name__ == "__main__":
    menu()
```
